```python
#Elabore um código que peça dois números e imprima a soma.
n1 = int(input("Digite um numero: "))
n2 = int(input("Digite outro numero: "))
soma = n1 + n2
print(soma)
```

```python
#Elabore um código que peça as 4 notas bimestrais e mostre a média
n1 = int(input("Digite uma nota: "))
n2 = int(input("Digite a segunda nota: "))
n3 = int(input("Digite a terceira nota: "))
n4 = int(input("Digite a ultima nota: "))
media = (n1 + n2 + n3 +n4) /4

print("sua media é: ", media)
```

```python
#Elabore um código que peça o raio de um círculo, calcule e mostre sua área
raio = int(input("digite o raio de um circulo: "))
area = (raio * raio) * 3.14

print("sua area é de: ", area)
```

```python
#Elabore um código que calcule a área de um quadrado, em seguida mostre o dobro desta área para o usuário.
lado = 10
area = lado * lado
dobro = area * 2

print("o dobro da sua area é: ", dobro)
```

```python
#Elabore um código que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês. Calcule e mostre o total do seu salário no referido mês em uma mensagem amigável.

sH = int(input("digite a qaunto você ganha por hora: "))
h = int(input("digite a quantia de horas trabalhadas por mes: "))
salario = sH * h

print("seu salario mensal é de: ", salario)
```

```python
#Elabore um código que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius.

F = float(input("digite a tempratura em F: "))
C = 5 * ((F-32) / 9)

print("a temperatura em C é: ", C)
```

```python
#Elabore um código que peça a temperatura em graus Celsius, transforme e mostre em graus Fahrenheit.
C = int(input("digite a tempratura em F: "))
F = C * 1.8 + 32

print("a temperatura em F é: ", F)
```
